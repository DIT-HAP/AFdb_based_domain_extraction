# 1.Load data

```python
AFdb_domain_file = Path(
    "../results/UP000002485_284812_SCHPO_v4_annotated_PAE_domain_based_on_pLDDT.tsv")

vDulab_domain_file = Path(
    "../results/vDulab_annotated_PAE_domain_based_on_pLDDT.tsv")

AFdb_domain = pd.read_csv(AFdb_domain_file, sep='\t')
vDulab_domain = pd.read_csv(vDulab_domain_file, sep='\t')

AFdb_domain["Source"] = "AFdb"
vDulab_domain["Source"] = "Dulab"

protein_info = pd.read_excel(
    "../data/0_raw/20240305_predicted_results/20230701_gene_info_with_protein_type.xlsx")

merged_domain = pd.merge(AFdb_domain, vDulab_domain, on=[
                         "Systematic ID", "uniprot_id", "domain_id"], suffixes=('_AFdb', '_vDulab'), how='outer').merge(protein_info[["gene_systematic_id", "Protein_type"]], left_on="Systematic ID", right_on="gene_systematic_id", how="left")

merged_domain.set_index(["Systematic ID", "uniprot_id"]).to_excel(
    "../results/AFdb_vDulab_domain_comparison.xlsx", index=True)
```

# 2.Update the protein domains

```python
updated_domains = AFdb_domain[~AFdb_domain["Systematic ID"].isin(
    vDulab_domain["Systematic ID"].unique())].copy()
updated_domains = pd.concat(
    [updated_domains, vDulab_domain[vDulab_domain["uniprot_id"].notna()]])
updated_domains.to_csv(
    "../results/UP000002485_284812_SCHPO_v4_and_vDulab_annotated_PAE_domain_based_on_pLDDT.tsv", sep='\t', index=False)
```
