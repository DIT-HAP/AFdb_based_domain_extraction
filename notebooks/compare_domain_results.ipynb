```python
import pandas as pd
```

```python
PAE_domains = pd.read_csv(
    "../results/UP000002485_284812_SCHPO_v4_reformated_PAE_domain_based_on_pLDDT.tsv",
    sep="\t",
    header=0,
).set_index(["uniprot_id", "Systematic ID", "protein_length", "domain_id"])
confident_domains_disordered_domains = pd.read_csv(
    "../results/UP000002485_284812_SCHPO_v4_confident_domains_based_on_pLDDT.tsv",
    sep="\t",
    header=0,
).set_index(["uniprot_id", "Systematic ID", "protein_length", "domain_id"])
```

```python
merged_domains = pd.concat(
    [PAE_domains, confident_domains_disordered_domains],
    axis=1,
    keys=["PAE_domain_version", "Confident_domain_version"],
)
```

```python
merged_domains.to_excel(
    "../results/compare_domain_results.xlsx", index=True, header=True
)
```
